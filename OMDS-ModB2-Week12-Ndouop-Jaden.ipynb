# 2. Weekly graph question

Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data?

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

```python
plt.plot(time_series_0)
plt.plot(time_series_1)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

Separate plots make it easy to see each week’s trend on its own without distractions, but they take more space and make direct comparison harder. Putting both weeks on the same plot saves space and makes differences easier to spot. The thing is it can also look cluttered or be confusing if the lines overlap or aren’t clearly labeled.

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 187-205 of the Storytelling With Data book as best you can. (Chapter Eight). You do not have to get the exact data values right, just the overall look and feel.

```python
import matplotlib.pyplot as plt
import numpy as np

# Fake data
depts = ["Sales","Marketing","HR","Finance","Product","Engineering","Support","Operations","Legal","IT"]
scores = np.array([78, 74, 71, 80, 86, 83, 68, 76, 79, 75])  # e.g., satisfaction (%)
order = np.argsort(scores)
depts = [depts[i] for i in order]
scores = scores[order]

overall = scores.mean()
highlight = "Support" 
highlight_idx = depts.index(highlight)

plt.figure(figsize=(8, 5))
ax = plt.gca()

for spine in ["top","right","left","bottom"]:
    ax.spines[spine].set_visible(False)
ax.set_yticks(range(len(depts)))
ax.set_yticklabels(depts)
ax.set_xticks([])  
ax.grid(axis="y", which="major", linewidth=0.4, alpha=0.2)

ax.axvline(overall, linestyle="--", linewidth=1, alpha=0.5)
ax.text(overall, len(depts)+0.2, f"Overall avg: {overall:.1f}", ha="center", va="bottom")

# Dots
ax.scatter(scores, range(len(depts)), s=35, alpha=0.7)
ax.scatter(scores[highlight_idx], highlight_idx, s=90, alpha=1.0)

for i, (s, d) in enumerate(zip(scores, depts)):
    ax.text(s + 0.6, i, f"{s}%", va="center", fontsize=9, alpha=0.9)

plt.title("Satisfaction by department (highlighting Support)", loc="left")
ax.text(0.0, 1.04, "Key point: Support trails the org average—opportunity to improve.",
        transform=ax.transAxes, ha="left", va="bottom")

plt.tight_layout()
plt.show()
```
